## Scryfall - Data Cleaning and Manipulation

Using unique artwork dataset downloaded from: https://scryfall.com/docs/api/bulk-data

```python
import pandas as pd
```

```python
df = pd.read_json('./data/scryfall/unique-artwork-20260108221918.json.zip', compression='zip')
df.head()
```

```python
print("Number of rows: ", len(df))
print("Column names:" , df.columns)
```

```python
# keep only needed columns
filtered_df = df[['id',
                'oracle_id',
                'name',
                'set_id',
                'set',
                'set_name',
                'artist_ids',
                'artist',
                'released_at',
                'type_line', # in  the format '[type] — [subtype]' (e.g. Creature — Kor Soldier)
                'keywords',
                'rarity',
                'color_identity',
                'cmc', # card's mana value
                'power',
                'toughness',
                'edhrec_rank',
                'legalities',
                'prices',
                'image_uris',
                'card_faces', # indicates whether card is multi-face
            ]]
filtered_df.head()
```

```python
# check for missing values
print("Count of null values in each column: ")
filtered_df.isnull().sum()
```

### Exclude multiface cards

Some magic cards are multi-faced, with faces divided on the front of the card like [split cards](https://scryfall.com/search?q=is%3Asplit) and [flip cards](https://scryfall.com/search?q=is%3Aflip), or double-sided like [transform cards](https://scryfall.com/search?q=is%3Atransform). These cards can have multiple artwork and/or ability (power/toughness). Scryfall indicates whether a card is multi-faced using the `card_faces` attribute, an array of card face objects with individual properties of name, type_line, power, toughness, image_uris, etc for each face. For single-faced cards, `card_faces` will be NaN. Based on the null values above, around 3.5k out of 50k cards from this dataset are multiface.

For our project, we will **exclude** multiface cards and only focus on cards with a single artwork/ability. We can expand to cover multiface cards later on.

```python
# keep only rows where card_faces IS null (these are single-faced cards)
cleaned_df = filtered_df[filtered_df['card_faces'].isnull()].copy()
cleaned_df.drop(columns=['card_faces'], inplace=True)
```

### Extract type & subtype from type_line

```python
# type_line is in the format '[type] — [subtype]' (e.g. Creature — Kor Soldier)
# note: not all cards have a subtype
cleaned_df['type'] = cleaned_df['type_line'].str.split('—').str[0].str.strip()
cleaned_df['subtype'] = cleaned_df['type_line'].str.split('—').str[1].str.strip()
cleaned_df.head()
```

### Extract legality, price (USD), and images from dictionary format

```python
# check formatting of legalities, prices, and image_uris dict
print(cleaned_df['legalities'][0])
print(cleaned_df['prices'][0])
print(cleaned_df['image_uris'][0])
```

```python
# use 'commander' and 'standard' play formats for now
cleaned_df['legality_commander'] = cleaned_df['legalities'].str.get('commander')
cleaned_df['legality_standard'] = cleaned_df['legalities'].str.get('standard')
cleaned_df['price_usd'] = cleaned_df['prices'].str.get('usd').astype(float)
# use 'normal' and 'art_crop' images for now
cleaned_df['image_uri_normal'] = cleaned_df['image_uris'].str.get('normal')
cleaned_df['image_uri_art_crop'] = cleaned_df['image_uris'].str.get('art_crop')

# check if any of these columns have null values
print("Count of null values in columns: ")
cleaned_df[['legality_commander', 'legality_standard', 'price_usd', 'image_uri_normal', 'image_uri_art_crop']].isnull().sum()
```

```python
# for cards missing USD price, try filling in with EUR price at 1 EUR = 1.20 USD conversion rate
cleaned_df['price_usd'] = cleaned_df['price_usd'].fillna(cleaned_df['prices'].str.get('eur').astype(float) * 1.2)

print("Number of cards with missing price (after EUR fill): ", cleaned_df['price_usd'].isnull().sum())
# drop rows with missing price
cleaned_df.dropna(subset=['price_usd'], inplace=True)
```

```python
# drop legalities, image_uris, and prices columns
final_df = cleaned_df.drop(columns=['legalities', 'image_uris', 'prices'])
```
